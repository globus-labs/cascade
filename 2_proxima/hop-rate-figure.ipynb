```python
import matplotlib.pyplot as plt
import numpy as np
import os

methods = ['DFT-only', 'Proxima', 'Proxima+Blend']
hop_rates = [2.68, 5.06, 2.80]
errors = [0.0834, 0.114, 0.0852]
colors = ['#222222', '#4472C4', '#C0392B']

fig, ax = plt.subplots(figsize=(6.25, 3.75))
# fig.patch.set_facecolor('white')
# ax.set_facecolor('#F5F5F7')

bars = ax.bar(methods, hop_rates, yerr=errors,
              color=colors, width=0.5,
              error_kw=dict(ecolor='#333333', elinewidth=1.5,
                            capsize=6, capth ick=1.5),
              zorder=3)

# DFT reference line
ax.axhline(y=2.68, color='#222222', linestyle='--',
           linewidth=1.5, alpha=0.6, zorder=2)

# Annotate bars with values
for bar, val, err in zip(bars, hop_rates, errors):
    ax.text(bar.get_x() + bar.get_width()/2,
            val + err + 0.1,
            f'{val:.2f}',
            ha='center', va='bottom',
            fontsize=16, fontweight='500',
            color='#222222')

# Proxima error annotation
ax.annotate('', xy=(1, 2.68), xytext=(1, 5.06),
            arrowprops=dict(arrowstyle='<->', color='#CC0000',
                           lw=1.5))
ax.text(1.3, 4.5, '~80% error', color='#CC0000',
        fontsize=16, fontweight='500', va='center')

# Proxima+Blend annotation
ax.text(2.75, 2 ,'~5% error',
        ha='center', color='#1D7A1D',
        fontsize=16, fontweight='500')

ax.set_ylabel('Hop rate (ps$^{-1}$)', fontsize=18)
ax.set_ylim(0, 6.5)
ax.set_xlim(-0.5, 2.8)
ax.tick_params(axis='both', labelsize=16)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#AAAAAA')
ax.spines['bottom'].set_color('#AAAAAA')
ax.yaxis.grid(True, color='white', linewidth=1.2, zorder=0)
ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig('./hop_rate_barplot.png',
            dpi=500, bbox_inches='tight',
            facecolor='white')
plt.show()
```
