## Look at the run parameters

```python
params = []
for d in run_dirs:
    try:
        params.append(json.loads((Path(d)/'params.json').read_text()))
    except: 
        print(f'couldnt find {d}/params.json; skipping')
        continue
    params[-1]['dir'] = d
params = pd.DataFrame(params)
```

```python
params = params.query('stress_tau == 25 and steps == 2048 and temperature == 800')
```

```python
params.columns
```

```python
param_vars = ['seed', 'temperature', 'n_blending_steps', 'min_target_frac', 'target_error', 'stress_tau', 'steps']
```

```python
_params = params[param_vars].sort_values(param_vars)
_params.describe()
```

```python
assert not _params.duplicated().any()
```

## Read all of the trajectory data into one dataframe

```python
data = []
hashes = []
for i, row in params.iterrows(): 
    # read in the results for each set of parameters
    run = row['dir']
    try:
        _data = pd.read_json(Path(run)/'proxima-log.json', lines=True)
    except: 
        print('couldnt read proximal log for', run)
        hashes.append(None)
        continue
    # make sure to have timestep
    _data.reset_index(names='timestep', inplace=True)

    # keep the hash of the run
    _hash = run.split('-')[-1]
    hashes.append(_hash)
    _data['hash'] = _hash 
    #traj = read(Path(run)/'md.traj', index=':')
    #break    
    # add the parameters to the experiment df for easy plotting, grouping, etc
    for var in param_vars:
        _data[var] = row[var]
    data.append(_data)

# track the hashes as well for param lookup
params['hash'] = hashes
data = pd.concat(data)
```

```python
# ── Compute per-run runtimes ──────────────────────────────────────────────────

# dft_only: min_target_frac == 1  (already defined above, but restated for clarity)
# experimental: min_target_frac < 1

# Sum step_time within each run (identified by 'hash') to get total wall-clock runtime
dft_runtime = (
    dft_only
    .groupby('hash')['step_time']
    .sum()
    .reset_index()
    .rename(columns={'step_time': 'runtime'})
)

exp_runtime = (
    experimental
    .groupby(['hash', 'min_target_frac', 'target_error', 'n_blending_steps'])['step_time']
    .sum()
    .reset_index()
    .rename(columns={'step_time': 'runtime'})
)

# Reference: median (or mean) runtime across all dft_only seeds
dft_ref_runtime = dft_runtime['runtime'].median()   # swap to .mean() if you prefer
print(f"DFT-only reference runtime: {dft_ref_runtime:.1f} s  (median over {len(dft_runtime)} runs)")

# ── Compute speedup = dft_ref / experimental_runtime ─────────────────────────
exp_runtime['speedup'] = dft_ref_runtime / exp_runtime['runtime']

# ── Plot: same facet style as the cost-accuracy figure ───────────────────────
g = sns.catplot(
    x='n_blending_steps',
    col='min_target_frac',
    hue='target_error',
    y='speedup',
    data=exp_runtime,
    kind='box',
    sharey=True,
    height=3,
    aspect=2 / (1 + np.sqrt(5)),
    palette=sns.color_palette("husl", exp_runtime['target_error'].nunique()),
)

# Reference line at speedup == 1 (DFT-only baseline)
for ax in g.axes.flat:
    ax.axhline(1, color='k', linestyle='--', linewidth=0.8, label='DFT-only')

g.set_axis_labels(r'$n_{blend}$', 'Speedup (×)')
g.set_titles(col_template=r'$f_{{target}}$ = {col_name}')
g.legend.set_title(r'$\epsilon_{target}$')
sns.move_legend(g, loc='center right', bbox_to_anchor=(1.18, 0.5))
plt.suptitle('Speedup of experimental runs over DFT-only baseline', y=1.02)
plt.tight_layout()
plt.savefig('speedup_npt.png', bbox_inches='tight', dpi=600)
plt.show()
```

```python
# ── Compute per-run runtimes and speedup ─────────────────────────────────────
dft_runtime = (
    dft_only[dft_only['timestep'] < 1023]
    .groupby('hash')['step_time']
    .sum()
    .reset_index()
    .rename(columns={'step_time': 'runtime'})
)

exp_runtime = (
    experimental
    .groupby(['hash', 'min_target_frac', 'target_error', 'n_blending_steps'])['step_time']
    .sum()
    .reset_index()
    .rename(columns={'step_time': 'runtime'})
)

dft_ref_runtime = dft_runtime['runtime'].median()
exp_runtime['speedup'] = dft_ref_runtime / exp_runtime['runtime']

# ── Layout parameters (match the FacetGrid above) ────────────────────────────
min_fracs   = sorted(cost_accuracy['min_target_frac'].unique())
target_errs = sorted(cost_accuracy['target_error'].unique())
blend_steps = sorted(cost_accuracy['n_blending_steps'].unique())
palette     = sns.color_palette("husl", len(target_errs))
err_to_color = dict(zip(target_errs, palette))

height  = 3                          # inches per cell (same as FacetGrid height=3)
aspect  = 2 / (1 + np.sqrt(5))
n_rows  = 3                          # rho error, frac_target, speedup
n_cols  = len(min_fracs)

fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(height * aspect * n_cols, height * n_rows),
    sharey='row',
)

# ── Rows 0 & 1: reproduce the original two metrics ───────────────────────────
metrics     = cost_accuracy['metric'].unique()          # e.g. ['rho_abs_pct_error', 'frac_target']
y_labels    = [r'% Absolute Error of $\rho$', r'% Timesteps with Target', r'Speedup (×)']

for row_idx, metric in enumerate(metrics):
    subset = cost_accuracy[cost_accuracy['metric'] == metric]
    for col_idx, mf in enumerate(min_fracs):
        ax = axes[row_idx, col_idx]
        col_data = subset[subset['min_target_frac'] == mf]
        sns.boxplot(
            x='n_blending_steps', y='value',
            hue='target_error',
            data=col_data,
            ax=ax,
            palette=err_to_color,
            order=blend_steps,
            hue_order=target_errs,
            legend=False,
        )
        ax.set_xlabel('')
        ax.set_ylabel(y_labels[row_idx] if col_idx == 0 else '')
        ax.set_title(f'min_target_frac = {mf}' if row_idx == 0 else '')

# ── Row 2: speedup ────────────────────────────────────────────────────────────
for col_idx, mf in enumerate(min_fracs):
    ax = axes[2, col_idx]
    col_data = exp_runtime[exp_runtime['min_target_frac'] == mf]
    sns.boxplot(
        x='n_blending_steps', y='speedup',
        hue='target_error',
        data=col_data,
        ax=ax,
        palette=err_to_color,
        order=blend_steps,
        hue_order=target_errs,
        legend=False,
    )
    ax.axhline(1, color='k', linestyle='--', linewidth=0.8)
    ax.set_xlabel(r'$n_{blend}$')
    ax.set_ylabel(r'Speedup (×)' if col_idx == 0 else '')
    ax.set_title('')

# ── Shared legend (mirrors the original SNS move_legend placement) ────────────
handles = [mpatches.Patch(color=err_to_color[e], label=e) for e in target_errs]
fig.legend(
    handles=handles,
    title=r'$\epsilon_{target}$',
    loc='center right',
    bbox_to_anchor=(1.1625, 0.5),
)

plt.tight_layout()
plt.savefig('cost_accuracy_speedup_npt.png', bbox_inches='tight', dpi=600)
plt.show()
```
