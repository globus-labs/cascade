# Make datasets for multihead replay

We are currently finetuning MACE using [multihead replay](https://mace-docs.readthedocs.io/en/latest/guide/multihead_finetuning.html).

The implementation uses ase.io.read to read an entire dataset into memory.

We may want to think about changing that, but for now I will just build a small dataset from the mace training set here to make that cheap to run in my test environment. 

```python
from glob import glob

from tqdm.notebook import tqdm
from ase.io import read, write
```

```python
training_files = glob('datasets/mace-mp/mptrj-gga-ggapu/*')
training_files[:10]
```

1000 is more than enough for just proof of concept runs

```python
N_KEEP = 1000
```

```python
sampled_data = []
for file in tqdm(training_files[:N_KEEP]):
    sampled_data.append(read(file))
```

```python
write(f'datasets/mace-mp/sampled_{N_KEEP}.traj', sampled_data)
```
